```python
question = [
    '비밀번호를 잊어버렸어요.',
'계정이 해킹당했어요.',
'로그인이 안 돼요.',
'이메일을 변경하고 싶어요.',
'2단계 인증이 안 돼요.'
]

# 비밀번호가 포함된 질문만 출력해주세요
keyword = "비밀번호"
for i, q in enumerate(question):
    if keyword in q:
        print(f"\n {i}, {q}")
```
